```python
import os
import glob
import fnmatch
import pandas as pd
import numpy as np
import librosa 
import librosa.display
import matplotlib.pyplot as plt
import IPython.display as ipd
import math
import tensorflow as tf
```

```python
data_dir = 'heartbeat_data'
```

```python
normal_data = data_dir + '/normal/'
murmur_data = data_dir + '/murmur/'
extrasystole_data = data_dir + '/extrasystole/'
extrasounds_data = data_dir + '/extrasounds/'
artifact_data = data_dir + '/extrasystole/'
test_data = data_dir + '/unlabeled/'
```

```python
print('Total Heart Auscultation Recordings:', len(os.listdir(normal_data)) 
                              + len(os.listdir(murmur_data))
                              + len(os.listdir(extrasystole_data))
                              + len(os.listdir(extrasounds_data))
                              + len(os.listdir(artifact_data)))
```

```python
x = np.array([len(os.listdir(normal_data)),
              len(os.listdir(murmur_data)),
              len(os.listdir(extrasystole_data)),
              len(os.listdir(extrasounds_data)),
              len(os.listdir(artifact_data))])
labels = ['Normal', 'Heart murmur', 'Extrasystole', 'Extra heart sounds', 'Artifact',]
plt.pie(x, labels = labels, autopct = '%.0f%%', radius= 1.5, textprops={'fontsize': 16})
plt.show()
```
